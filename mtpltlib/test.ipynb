# MatplotLib and NumPy

```python
import matplotlib.pyplot as plt
import numpy as np

x = np.array([0, 1, 2, 3, 4, 5])
y = np.array([0, 1, 4, 9, 16, 25])
z = np.array([0, 1, 8, 27, 64, 125])

lnStyle = dict(linestyle = 'dashed', marker='o', color='g', markersize=8)

# plt.plot(x, y, marker='o', linestyle='dashdot', color='b', markersize=8)
plt.plot(x, y, **lnStyle)
plt.title('Quadratic Function')
plt.xlabel('x')
plt.ylabel('y')

plt.plot(x, z, **lnStyle)
plt.title('Cubic Function')
plt.xlabel('x')
plt.ylabel('z')
plt.grid()
```

# Bar Charts

```python
import matplotlib.pyplot as plt

food = ['Pizza', 'Burger', 'Pasta', 'Salad']
calories = [300, 500, 400, 150]

plt.bar(food, calories, color=['red', 'blue', 'green', 'orange'])
plt.title('Calories in Different Foods', fontfamily='monospace')
plt.xlabel('Food')
plt.ylabel('Calories')
plt.grid(alpha=0.5)
```

# Pie Chart

```python
import matplotlib.pyplot as plt
import numpy as np

food = ['Pizza', 'Burger', 'Pasta', 'Salad']
calories = [300, 500, 400, 150]

plt.pie(calories, labels=food, 
                autopct='%1.2f%%',
                explode=[0, 0.1, 0, 0],
                shadow=True)
plt.title('Calories Distribution', fontfamily='monospace')
plt.axis('equal')
plt.show()
```

# Scatter Chart

```python
import matplotlib.pyplot as plt
import numpy as np

x = np.array([0, 1, 2, 3, 4, 5])
y = np.array([0, 1, 4, 9, 16, 25])
z = np.array([0, 1, 8, 27, 64, 125])

plt.scatter(x, y, label='Quadratic')
plt.scatter(x, z, label='Cubic')
plt.title('Scatter Plot of Functions')
plt.legend()

plt.hist(x)
```

# Linear Regression

```python
from sklearn.linear_model import LinearRegression
import numpy as np

# study hours
x = np.array([[1], [2], [3], [4]])

# marks
y = np.array([10,20,30,40])

model = LinearRegression()

model.fit(x,y)

prediction = model.predict([[5]])

print(prediction)
```
