```python
import pandas as pd
import numpy as np

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler


# ============================================================
# 1. Load Dataset
# ============================================================

housing = pd.read_csv(
    "~/MyCodes/MachineLearning/Book/datasets/housing/housing.csv"
)


# ============================================================
# 2. Separate Features (X) and Target (y)
# ============================================================

X = housing.drop("median_house_value", axis=1)

y = housing["median_house_value"]


# ============================================================
# 3. Select Numerical Features
# ============================================================

X_num = X.select_dtypes(include=np.number)
print(X_num)


# ============================================================
# 4. Train-Test Split
# ============================================================

X_train, X_test, y_train, y_test = train_test_split(
    X_num,
    y,
    test_size=0.2,
    random_state=42
)


# ============================================================
# 5. Handle Missing Values
# ============================================================
# log1p() cannot handle NaN values.
# So fill missing numerical values using training median.

train_medians = X_train.median()

X_train = X_train.fillna(train_medians)
X_test = X_test.fillna(train_medians)


# ============================================================
# 6. Log Transformation
# ============================================================
# These features are strongly right-skewed / heavy-tailed.

heavy_tail_features = [
    "total_rooms",
    "total_bedrooms",
    "population",
    "households"
]

for column in heavy_tail_features:
    X_train[column] = np.log1p(X_train[column])
    X_test[column] = np.log1p(X_test[column])


# ============================================================
# 7. Standardization                                          
# ============================================================ 

scaler = StandardScaler()

X_train_scaled = scaler.fit_transform(X_train)

X_test_scaled = scaler.transform(X_test)


# ============================================================
# 8. Convert NumPy Arrays back to DataFrame
# ============================================================

X_train_scaled = pd.DataFrame(
    X_train_scaled,
    columns=X_train.columns,
    index=X_train.index
)

X_test_scaled = pd.DataFrame(
    X_test_scaled,
    columns=X_test.columns,
    index=X_test.index
)


# ============================================================
# 9. Check the Result
# ============================================================

# def sep():
#     print("-" * 64)


# print("Original Training Data:")
# print(X_train.head())

# sep()

# print("Scaled Training Data:")
# print(X_train_scaled.head())

# sep()

# print("Mean after scaling:")
# print(X_train_scaled.mean())

# sep()

# print("Standard deviation after scaling:")
# print(X_train_scaled.std())

# sep()
```
