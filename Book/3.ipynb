```python
import numpy as np 
from zlib import crc32
import pandas as pd

def is_id_in_test_set(idx, ratio):
    return crc32(np.int64(idx)) < (2 ** 32) * ratio;

def split_data_with_id_hash(df, ratio, id_column):
    in_test_set = []

    for idx in df[id_column]:
        flag = is_id_in_test_set(idx, ratio)
        in_test_set.append(flag)

    in_test_set_series = pd.Series(in_test_set, index=df.index)
    return df.loc[~in_test_set_series], df.loc[in_test_set_series]

raw_data = {
        'id': [101, 102, 103, 104, 105],
        'median_house_value': [450000, 200000, 150000, 300000, 500000]
        }

df = pd.DataFrame(raw_data)

train_set, test_set = split_data_with_id_hash(df, 0.4, 'id')

print(f"Train set \n")
print(train_set)
print(f"Test set \n")
print(test_set)
```
