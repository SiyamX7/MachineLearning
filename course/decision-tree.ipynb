```python
import pandas as pd 
import numpy as np 
import matplotlib.pyplot as plt 

dataset = pd.read_csv('/home/siyam/Desktop/Dataset/Part 2 - Regression/Data/Position_Salaries.csv')
x = dataset.iloc[:, 1:-1].values 
y = dataset.iloc[:, -1].values

# Training Decision Tree regression model
from sklearn.tree import DecisionTreeRegressor
regressor = DecisionTreeRegressor(random_state=1)
regressor.fit(x,y)
value = float(input("enter your clients level:"))
regressor.predict([[value]])
```

```python
plt.scatter(x,y, color='r')
plt.plot(x, regressor.predict(x), color ='b')
```

```python
x_grid = np.arange(x.min(), x.max(), 0.0001).reshape(-1,1)
plt.scatter(x, y, color='r')
plt.plot(x_grid, regressor.predict(x_grid), color='b')
plt.xlabel("level")
plt.ylabel("salary")
plt.title("Decision tree regression")
plt.legend(['actual', 'predicted'])
```
