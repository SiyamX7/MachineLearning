```python
import pandas as pd 
import matplotlib.pyplot as plt 
import numpy as np 

dataset = pd.read_csv('/home/siyam/Desktop/Dataset/Part 2 - Regression/Data/Position_Salaries.csv')
x = dataset.iloc[:, 1:-1].values
y = dataset.iloc[:, -1].values

from sklearn.preprocessing import StandardScaler
sc_x = StandardScaler()
sc_y = StandardScaler()

scaled_x = sc_x.fit_transform(x)

scaled_y = sc_y.fit_transform(y.reshape(-1,1))

# print(scaled_y)

# print(x)
# print()
# print(y)

from sklearn.svm import SVR
regressor = SVR(kernel='rbf')

# print(y)

regressor.fit(scaled_x, scaled_y.ravel())

# g = float(input("Enter the level: "))
g = 4.353
value = regressor.predict(sc_x.transform([[g]])).reshape(-1,1)

print(value)
print(sc_y.inverse_transform(value))


plt.scatter(x,y,color='r')
plt.plot(x, sc_y.inverse_transform(regressor.predict(scaled_x).reshape(-1,1)) , color= 'b')
plt.title('SVR Regression')
plt.xlabel('Position Level')
plt.ylabel('Salary')
plt.legend(['Actual', 'Predicted'])
```

```python
x_grid = np.arange(x.min(), x.max(), 0.1).reshape(-1,1)
plt.scatter(x, y, color = 'r')
plt.plot(x_grid, sc_y.inverse_transform(regressor.predict(sc_x.transform(x_grid)).reshape(-1,1)))
plt.title("SVR Prediction")
plt.xlabel("level")
plt.ylabel("SALARY")
plt.legend(['actual', 'predicted', 'gg'])
```
